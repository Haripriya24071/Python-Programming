3. Largest of Three Numbers

```python
a = 15
b = 40
c = 25

if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print("Largest number is:", largest)
```

4. Even or Odd

```python
number = 17
if number % 2 == 0:
    print(number, "is Even")
else:
    print(number, "is Odd")
```

5. Multiplication Table

```python
number = 7
print(f"Multiplication table of {number}:")
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

6. Factorial

```python
number = 5
factorial = 1
for i in range(1, number + 1):
    factorial *= i
print(f"Factorial of {number} is:", factorial)
```

7. Prime Number

```python
number = 29
is_prime = True
if number <= 1:
    is_prime = False
else:
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            is_prime = False
            break

if is_prime:
    print(number, "is a Prime Number")
else:
    print(number, "is Not a Prime Number")
```

8. Reverse a String

9. Sum of List Elements

10. Fibonacci Series

```python
n = 10
a, b = 0, 1
print(f"First {n} terms of Fibonacci series:")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
print()
```
