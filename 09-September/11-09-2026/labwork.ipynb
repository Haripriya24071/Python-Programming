1. Find Smallest Number in a List

```python
nums = [12, 45, 2, 67, 34]
print("Numbers list:", nums)
print("Smallest number is:", min(nums))
```

2. Find Average of Numbers in a List

```python
nums = [10, 20, 30, 40, 50]
avg = sum(nums) / len(nums)
print("Numbers list:", nums)
print("Average is:", avg)
```

3. Count Occurrences of a Number in a List

```python
nums = [1, 2, 3, 2, 4, 2, 5]
target = 2
count = nums.count(target)
print("Numbers list:", nums)
print(f"Count of {target} is:", count)
```

4. Reverse a List

```python
nums = [1, 2, 3, 4, 5]
print("Original list:", nums)
nums.reverse()
print("Reversed list:", nums)
```

5. Sort a List in Ascending Order

```python
nums = [45, 12, 89, 3, 22]
print("Original list:", nums)
nums.sort()
print("Sorted list:", nums)
```

6. Common Elements of Two Tuples

```python
t1 = (1, 2, 3, 4)
t2 = (3, 4, 5, 6)
common = tuple(set(t1) & set(t2))
print("Tuple 1:", t1)
print("Tuple 2:", t2)
print("Common elements:", common)
```

7. Convert List to Tuple

```python
my_list = [1, 2, 3, 4, 5]
my_tuple = tuple(my_list)
print("List:", my_list)
print("Converted tuple:", my_tuple)
```

8. Convert Tuple to List

```python
my_tuple = (4, 5, 6, 7, 8)
my_list = list(my_tuple)
print("Tuple:", my_tuple)
print("Converted list:", my_list)
```

9. Set Union, Intersection, and Difference

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
print("Set 1:", s1)
print("Set 2:", s2)
print("Union:", s1 | s2)
print("Intersection:", s1 & s2)
print("Difference (s1 - s2):", s1 - s2)
```

10. Remove Duplicate Values from List Using Set

```python
nums = [1, 2, 2, 3, 4, 4, 5]
unique_nums = list(set(nums))
print("Original list:", nums)
print("List without duplicates:", unique_nums)
```
