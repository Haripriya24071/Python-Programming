1. Multiplication Table

```python
num = 5
print(f"Multiplication table of {num}:")
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

2. Right Triangle Star Pattern

```python
rows = 5
for i in range(1, rows + 1):
    print("*" * i)
```

3. Inverted Triangle Star Pattern

```python
rows = 5
for i in range(rows, 0, -1):
    print("*" * i)
```

4. Pyramid Star Pattern

```python
rows = 5
for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
```

5. Fibonacci Series

```python
n = 10
a, b = 0, 1
print(f"Fibonacci series (first {n} terms):")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
print()
```
